# Brincando com as variáveis
Vamos modelar um problema bobo de causalidade do 0, simulando dados da vida real. Queremos saber se dormir 8h antes de uma prova e dormir 4h antes de uma prova afetam em algo o resultado da prova, temos então:
- Unidades ($U$): Estudantes
- Causa de interesse (t v.s c) ~ Dormir 8h v.s dormir 4h
- Resposta ($Y$): Nota no exame (0-100)
- Variável de confusão: Ansiedade do estudante (0-10)

```python
import numpy as np

n = 1000  # Quantidade de alunos
rng = np.random.default_rng(seed=42)  # Gerador de números aleatórios com semente fixa
```

Primeiro, modelamos a ansiedade. Aqui, vamos modelar ela como $\approx \cal{N}(5, 4)$, mas, na vida real, pode acontecer de não conseguirmos medir essa variável

```python
ansiedade = rng.normal(loc=5, scale=2, size=n).clip(0,10)  # Nível de ansiedade (média 5, desvio padrão 2)
```

Agora modelamos a probabilidade do estudante dormir pouco conforme fica mais ansioso, de modo que aumentar a ansiedade aumenta a chance dele dormir pouco

```python
prob_dormir_pouco = 1 / (1 + np.exp(-(ansiedade - 5) * 0.6))
dormiu_pouco = rng.binomial(1, prob_dormir_pouco, size=n)  # 1 = 4h, 0 = 8h
```
